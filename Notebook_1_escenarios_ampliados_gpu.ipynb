## 3.3 Verificación de valores atípicos

#### **Metodología**

Para la detección de valores atípicos se aplicó un enfoque adaptativo basado en la distribución de cada variable numérica. Primero, se seleccionaron todas las variables de tipo `int64` y `float64`. Posteriormente, se calculó la **asimetría (skewness)** de cada variable con el fin de determinar el método de detección más adecuado.

- **Distribuciones aproximadamente simétricas (|skew| < 1):**  
  Se utilizó el método de **Z-score**, considerando como atípicos los valores con |z| > 3.

- **Distribuciones asimétricas (|skew| ≥ 1):**  
  Se aplicó el método del **Rango Intercuartílico (IQR)**, identificando como atípicos los valores fuera del intervalo:  
  Q1 − 1.5·IQR  y  Q3 + 1.5·IQR.

Para cada variable se calculó la cantidad y el porcentaje de observaciones atípicas, permitiendo evaluar su impacto real en la base de datos.

```python
numericas = base_final.select_dtypes(include=["int64", "float64"])

reporte_outliers = []

for col in numericas.columns:

    serie = numericas[col].dropna()

    if len(serie) < 30:
        continue

    asimetria = skew(serie)

    # Si la distribución es aproximadamente simétrica
    if abs(asimetria) < 1:
        # Usamos método Z-score
        z_scores = (serie - serie.mean()) / serie.std()
        outliers = np.sum(np.abs(z_scores) > 3)
        metodo = "Z-score"

    # Si es asimétrica
    else:
        # Usamos IQR
        Q1 = serie.quantile(0.25)
        Q3 = serie.quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = np.sum((serie < lower) | (serie > upper))
        metodo = "IQR (asimétrica)"

    reporte_outliers.append({
        "Variable": col,
        "Asimetria": round(asimetria, 3),
        "Metodo_usado": metodo,
        "Cantidad_outliers": outliers,
        "Porcentaje_outliers": round((outliers / len(serie)) * 100, 2)
    })

reporte_outliers = pd.DataFrame(reporte_outliers)

reporte_outliers.sort_values("Porcentaje_outliers", ascending=False).head(15)
```

### Resultados

Los hallazgos muestran que:

- En la mayoría de los **puntajes académicos**, el porcentaje de valores atípicos es bajo (generalmente < 2%).
- En variables clave como `punt_global`, `punt_matematicas`, `punt_lectura_critica` y `punt_ingles`, los valores extremos representan menos del 1%.
- Algunas variables como `cole_codigo_icfes` e `inst_cod_institucion` presentan mayor proporción de atípicos; sin embargo, esto se debe a que son **códigos identificadores**, no variables métricas continuas.

### **Justificación de la Decisión**

Se decidió **no eliminar los valores atípicos** por las siguientes razones:

1. **Baja proporción relativa**, sin impacto significativo en la estructura global.
2. **Plausibilidad sustantiva**, ya que en pruebas estandarizadas los valores extremos son esperables.
3. **Naturaleza de ciertas variables**, donde la detección de outliers no tiene relevancia analítica.

No se evidencian anomalías sistemáticas ni errores masivos en los datos. Los valores extremos identificados son coherentes con la naturaleza educativa del fenómeno analizado. Por lo tanto, se optó por conservar la integridad de la información para evitar sesgos en análisis posteriores.

Ahora para poder terminar el tratamiento de datos nulos se va a realizar pruebas de sesgo de los datos si los datos tienen un coeficiente de asimetría entre -0.5 y 0.5 se considera simétrica y se imputará la media o la moda en caso de los datos categóricos. Si el coeficente de asimetría se sale de este rango entonces se imputara con la mediana para no sesgar los datos.

```python
# Crear reporte
reporte_imputacion = []

# Iterar sobre todas las columnas
for col in df.columns:

    # Solo procesar si tiene nulos
    if df[col].isnull().sum() > 0:

        # Verificar si es numérica
        if pd.api.types.is_numeric_dtype(df[col]):

            skewness = df[col].skew()

            # Si es simétrica → imputar con media
            if -0.5 <= skewness <= 0.5:

                valor_imputacion = df[col].mean()
                metodo = "media"

            # Si es asimétrica → imputar con mediana
            else:

                valor_imputacion = df[col].median()
                metodo = "mediana"

            df[col].fillna(valor_imputacion, inplace=True)

            reporte_imputacion.append({
                "columna": col,
                "tipo": "numérica",
                "skewness": skewness,
                "metodo": metodo,
                "nulos_imputados": df[col].isnull().sum()
            })

        # Si es categórica → imputar con moda
        else:

            moda = df[col].mode()

            if len(moda) > 0:
                valor_imputacion = moda[0]
            else:
                valor_imputacion = "No aplica"

            df[col].fillna(valor_imputacion, inplace=True)

            reporte_imputacion.append({
                "columna": col,
                "tipo": "categórica",
                "skewness": "No aplica",
                "metodo": "moda",
                "nulos_imputados": df[col].isnull().sum()
            })

# Convertir reporte a DataFrame
reporte_imputacion = pd.DataFrame(reporte_imputacion)

print("Imputación completada.")
print("\nResumen:")
print(reporte_imputacion)
```

## 4.4 Tratamiento Columnas desbalanceadas
En este apartado se realiza la identificación y el tratamiento de las columnas desbalanceadas pus estas pueden generar un sesgo a la hora de hacer cualquier tipo e análisis. Una variable se considera desbalanceada cuando una categoría representa más del 80% de las observaciones, altamente desbalanceada cuando supera el 90% y extremadamente desbalanceada cuando supera el 95%, ya que esto puede afectar la capacidad de los modelos estadísticos para identificar patrones en las categorías minoritarias (He & Garcia, 2009; Kuhn & Johnson, 2013).

```python
# Crear copia
df_prebalanceado = df.copy()

# Umbral de desbalance
umbral = 0.80

# Guardar reporte
columnas_desbalanceadas = []

for col in df.select_dtypes(include=['object', 'category']).columns:

    distribucion = df[col].value_counts(normalize=True)
    max_prop = distribucion.max()

    if max_prop > umbral:

        columnas_desbalanceadas.append({
            "columna": col,
            "categoria_dominante": distribucion.idxmax(),
            "porcentaje": round(max_prop*100, 2),
            "numero_categorias": df[col].nunique()
        })

# Convertir a DataFrame
reporte_desbalance = pd.DataFrame(columnas_desbalanceadas)

print("Columnas desbalanceadas:")
print(reporte_desbalance)
```

```python
cols_nse = [
    "estu_nse_establecimiento",
    "estu_nse_individual_saber11",
    "estu_nse_individual_saberpro",
    # "AQUI_TU_CUARTA_COLUMNA"  # si existe, agrégala
]

orden_nse = ["NSE1", "NSE2", "NSE3", "NSE4"]

def normaliza_nse_imputa(x):
    # Imputación por defecto
    default = "NSE2"

    if pd.isna(x):
        return default

    s = str(x).strip().upper()

    # Si ya viene bien
    if s in orden_nse:
        return s

    # Si viene como '1','2','3','4' o '1.0' etc.
    try:
        n = int(float(s))
        if n in [1, 2, 3, 4]:
            return f"NSE{n}"
        else:
            return default
    except:
        return default

for c in cols_nse:
    if c in df.columns:
        df[c] = df[c].apply(normaliza_nse_imputa)
        # opcional: dejarla como categórica ordenada
        df[c] = pd.Categorical(df[c], categories=orden_nse, ordered=True)
    else:
        print(f"⚠️ No existe en df: {c}")
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_cluster)
```

```python
# KMeans con GPU opcional (cuML) y fallback a sklearn
import numpy as np

try:
    from cuml.cluster import KMeans as AcceleratedKMeans
    _kmeans_backend = "GPU (cuML)"
except Exception:
    from sklearn.cluster import KMeans as AcceleratedKMeans
    _kmeans_backend = "CPU fallback (sklearn)"

def _kmeans_to_numpy(values):
    if hasattr(values, "to_numpy"):
        values = values.to_numpy()
    elif hasattr(values, "get"):
        values = values.get()
    return np.asarray(values)

print(f"Backend KMeans activo: {_kmeans_backend}")

inercias = []

for k in range(2, 10):
    kmeans = AcceleratedKMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    inertia = getattr(kmeans, "inertia_", None)
    if inertia is None and hasattr(kmeans, "score"):
        inertia = -kmeans.score(df_scaled)
    inercias.append(float(_kmeans_to_numpy(inertia)))

plt.plot(range(2, 10), inercias, marker='o')
plt.xlabel('Número de clusters')
plt.ylabel('Inercia')
plt.title(f'Método del codo ({_kmeans_backend})')
plt.show()

```

# Modelos de machine learning

```python
import re

def _cols_like(df, patterns):
    """Devuelve columnas que matchean cualquiera de los patrones (regex) en patterns."""
    cols = []
    for c in df.columns:
        for p in patterns:
            if re.search(p, str(c), flags=re.IGNORECASE):
                cols.append(c)
                break
    return cols


def build_df_saber11(
    df,
    ignore_saber11=None,
    keep_targets=("nivel_saber11", "punt_global_saber11"),
    drop_percentiles=True,
    drop_desempenos=True,
    drop_puntajes_competencias=True,
):
    """
    Saber11:
    - Quita TODO SaberPro
    - Quita niveles de desempeño, puntajes por competencia y percentiles de Saber11
    - Mantiene targets indicados en keep_targets (si existen)
    """
    ignore_saber11 = set(ignore_saber11 or [])

    df_out = df.copy()

    # 0) Arreglo por si hay columnas vacías tipo ""
    if "" in df_out.columns:
        df_out = df_out.drop(columns=[""], errors="ignore")

    # 1) Quitar todo lo de SaberPro (por sufijos/patrones típicos)
    #    Ajusta patrones si tienes otras convenciones
    drop_saberpro = set(_cols_like(df_out, [
        r"_saberpro\b",            # columnas que terminan/contienen _saberpro
        r"\bsaberpro\b",           # cualquier mención saberpro
        r"^mod_",                  # módulos saber pro (mod_*)
        r"^percentil_.*saberpro",  # percentiles marcados explícitamente
        r"^punt_global\b",         # OJO: esto quita punt_global (genérico). Lo manejo más abajo si es target.
        r"^percentil_global\b",    # percentil_global típico pro
        r"^percentil_nbc\b",       # percentil_nbc típico pro
    ]))

    # 2) Quitar niveles/puntajes/percentiles de Saber11 (según tu solicitud)
    drop_saber11_perf = set()
    if drop_desempenos:
        drop_saber11_perf |= set(_cols_like(df_out, [r"^desemp_"]))

    if drop_puntajes_competencias:
        drop_saber11_perf |= set(_cols_like(df_out, [r"^punt_", r"^recaf_punt_"]))

    if drop_percentiles:
        drop_saber11_perf |= set(_cols_like(df_out, [r"^percentil_"]))

    # 3) Quitar también lo que tengas en ignore_saber11 explícito
    drop_explicit = set([c for c in ignore_saber11 if c in df_out.columns])

    # 4) Pero CONSERVAR targets si existen
    keep_targets = [t for t in keep_targets if t in df_out.columns]

    drop_all = (drop_saberpro | drop_saber11_perf | drop_explicit) - set(keep_targets)

    df_out = df_out.drop(columns=sorted(drop_all), errors="ignore")

    return df_out


def build_df_saberpro(
    df,
    ignore_saberpro=None,
    keep_targets=("nivel_saberpro", "punt_global_saberpro"),
    drop_modulos=True,
    drop_percentiles_saberpro=True,
    drop_niveles_modulos=True,
    drop_puntajes_modulos=True,
    drop_global_generic=False,   # si True, elimina punt_global genérico (si no es target)
):
    """
    SaberPro:
    - Deja TODO lo de Saber11 (contexto) (NO lo elimina)
    - Quita de SaberPro: puntajes/niveles/percentiles de módulos
    - Mantiene targets indicados en keep_targets (si existen)
    """
    ignore_saberpro = set(ignore_saberpro or [])

    df_out = df.copy()

    if "" in df_out.columns:
        df_out = df_out.drop(columns=[""], errors="ignore")

    drop_saberpro_perf = set()

    # 1) Quitar módulos pro (normalmente aquí está TODO: punt, pnal/pnbc, desem, etc.)
    if drop_modulos:
        drop_saberpro_perf |= set(_cols_like(df_out, [r"^mod_"]))

    # 2) Quitar percentiles pro (incluye percentil_global, percentil_nbc y percentil_*_saberpro)
    if drop_percentiles_saberpro:
        drop_saberpro_perf |= set(_cols_like(df_out, [
            r"^percentil_global\b",
            r"^percentil_nbc\b",
            r"^percentil_.*saberpro\b",
            ]))

    # 3) Si quieres ser más explícito en “niveles”/“puntajes” (ya están dentro de mod_ normalmente)
    if drop_niveles_modulos:
        drop_saberpro_perf |= set(_cols_like(df_out, [r"_desem\b", r"_desem_"]))

    if drop_puntajes_modulos:
        drop_saberpro_perf |= set(_cols_like(df_out, [r"_punt\b", r"_punt_"]))

    # 4) Quitar ignore_saberpro explícito
    drop_explicit = set([c for c in ignore_saberpro if c in df_out.columns])

    # 5) Manejo de punt_global genérico (solo si no es target)
    if drop_global_generic:
        drop_saberpro_perf |= set(_cols_like(df_out, [r"^punt_global\b"]))

    # 6) Mantener targets
    keep_targets = [t for t in keep_targets if t in df_out.columns]
    drop_all = (drop_saberpro_perf | drop_explicit) - set(keep_targets)

    df_out = df_out.drop(columns=sorted(drop_all), errors="ignore")

    return df_out


# ====== USO ======
# df = tu dataframe "mezclado" (con columnas de saber11 y saberpro)

df_saber11 = build_df_saber11(
    df,
    ignore_saber11=ignore_saber11,
    keep_targets=("nivel_saber11", "punt_global_saber11"),
)

# NOTA IMPORTANTE:
# En df_saberpro yo dejé un patrón fuerte que puede borrar percentiles SABER11 también
# por el regex r"^percentil_.*\b". Si tú SÍ quieres conservar percentiles de SABER11 en df_saberpro,
# cambia drop_percentiles_saberpro para que SOLO borre los de SaberPro (los que tengan _saberpro o percentil_global/nbc).
df_saberpro = build_df_saberpro(
    df,
    ignore_saberpro=ignore_saberpro,
    keep_targets=("nivel_saberpro", "punt_global_saberpro"),
    drop_percentiles_saberpro=True,
    drop_global_generic=False,
)

print("df original:", df.shape)
print("df_saber11:", df_saber11.shape)
print("df_saberpro:", df_saberpro.shape)
```

## Predictivos

En esta sección se ejecutan los escenarios 1 a 6 solo con los modelos predictivos que tienen implementación GPU activa en el notebook.  
El criterio de selección es **MAPE** y en el reporte final se muestran **MAE, RMSE, $R^2$ y MAPE** tanto en prueba como en validación cruzada.



```python
# =========================
# Predictivos - ejecución de escenarios 1 a 6
# =========================
regression_scenarios = [s for s in SCENARIOS if s["escenario"] <= 6]

resultados_regresion = {
    "saber11": {},
    "saberpro": {},
}

for scenario in regression_scenarios:
    resultados_regresion["saber11"][scenario["escenario"]] = evaluate_regression_models(
        df_model=df_saber11_pr,
        target_col="punt_global_saber11",
        ordinal_specs=scenario["ordinal_specs"],
        scale_mode=scenario["scale_mode"],
        top_n=5,
    )
    resultados_regresion["saberpro"][scenario["escenario"]] = evaluate_regression_models(
        df_model=df_saberpro_pr,
        target_col="punt_global_saberpro",
        ordinal_specs=scenario["ordinal_specs"],
        scale_mode=scenario["scale_mode"],
        top_n=5,
    )

resumen_reg_saber11 = pd.concat(
    [
        resultados_regresion["saber11"][esc]["tuned_summary"].assign(Escenario=esc)
        for esc in resultados_regresion["saber11"]
    ],
    ignore_index=True,
)
resumen_reg_saberpro = pd.concat(
    [
        resultados_regresion["saberpro"][esc]["tuned_summary"].assign(Escenario=esc)
        for esc in resultados_regresion["saberpro"]
    ],
    ignore_index=True,
)

print("Resumen predictivo - Saber 11")
display(resumen_reg_saber11.sort_values(by=["Escenario", "Test_MAPE_%"]).reset_index(drop=True))

print("Resumen predictivo - Saber Pro")
display(resumen_reg_saberpro.sort_values(by=["Escenario", "Test_MAPE_%"]).reset_index(drop=True))

# Ejemplo de consulta detallada:
# resultados_regresion["saber11"][1]["tuned_results"]["RandomForestRegressor"]["cv_folds"]
# resultados_regresion["saber11"][1]["tuned_results"]["RandomForestRegressor"]["cv_summary"]
```
